```python
a = 'Anas' # single quoted string 
b = " Anas " # double quoted string
c = '''Anas''' # Triple quoted string
```

```python
name = "Anas" 
nameShort = name[0:3] # start from index 0 all the way till 3 ( excluding 3)
print(nameShort)
character1 = name[1]
print(character1)

# Negative slicing
name = "Anas"

print(name[-4: -1])
print(name[1:4])

print(name[:4]) # is same as print(name[0:4])
print(name[1:]) # is same as print(name[1:5])
print(name[1:5])
```
