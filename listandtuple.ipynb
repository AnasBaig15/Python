Lists

```python
friends = ["Apple","Orange",5,345.06,False,"Ahmed","Salaar"]
print(friends[0])
friends[0] = "Grapes" # unlike Strings lists are mutable
print(friends[0])
print(friends[1:4]) # prints from index 1 to 3

friends.append("Shayan") # adds an element to the end of the list
print(friends)

l1 = [1,2,10,4,5,6,7,8,9,3]
l1.sort()
l1.reverse()
l1.insert(5,3333) # insert 3333 at index 5

value = l1.pop(3) # removes the element at index 3 and returns it
```
