```python
for i in range(100):
    if(i == 34):
        break # Exit the loop right now
    # print(i)

for i in range(100):
    if(i == 34):
        continue # Skip this iteration
    # print(i)

for i in range(645):
    pass # skipping method 

i= 0
while(i<45):
    print(i)
    i += 1
```
