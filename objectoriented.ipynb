Object Oriented Programming

```python
class Employee:
    language = "Py"
    salary = 100000

anas = Employee()
print(anas.salary,anas.language)
```

```python
# Instance Attribute

class Employee:
    language = "python"
    salary = 10000
anas = Employee()
anas.language = "JavaScript" # This is instance attribute
print(anas.language, anas.salary)

```

```python
# Self Attribute

class Employee:
    language = "python"
    salary = 10000
    def getInfo(self):
        print(f"The language is {self.language} . The salary is {self.salary}")
    def greet():
        print("Good morning")    

anas = Employee()
#anas.language = "JavaScript" # This is instance attribute
anas.getInfo()
Employee.getInfo(anas)
```

```python
# Static Method
@staticmethod
def greet():
    print("Good Morning")
```

```python
class Employee:
    language = "python"
    salary = 10000

    def __init__(self,name , salary language): # dunder method which is automatically called
        self.name = name
        self.salary = salary
        self.language = language
        print("I am creating an object")
        
    def getInfo(self):
        print(f"The language is {self.language} . The salary is {self.salary}")

    @staticmethod
    def greet():
        print("Good Morning")

anas = Employee("Anas",10000," Javascript")
anas.name = "anas"
print(anas.name, anas.salary)
```
